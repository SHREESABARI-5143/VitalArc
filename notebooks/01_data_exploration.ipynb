Analyze dataset distributions, class balance, image dimensions, and anterior segment color channel properties.

```python
import os
import glob
import json
from collections import Counter
from PIL import Image

# Load sample metadata
metadata_path = "../data/sample/sample_metadata.json"
if os.path.exists(metadata_path):
    with open(metadata_path, 'r') as f:
        meta = json.load(f)
    print(f"Loaded {len(meta['samples'])} sample patient records.")
    for sample in meta['samples']:
        print(f" - [{sample['ground_truth']}] {sample['id']}: {sample['findings']}")
else:
    print("Sample metadata file found at expected path.")
```
